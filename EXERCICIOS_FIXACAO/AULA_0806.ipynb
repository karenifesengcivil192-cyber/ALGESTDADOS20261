```python
# 1) Validação de Limite Numérico - LISTA 08/06

def verificar(numero):
    return numero > 10

numero = int(input("Digite um número inteiro: "))

if verificar(numero):
    print("O número é maior que 10.")
else:
    print("O número não é maior que 10.")
```

```python
# 2) Classificação de Paridade - LISTA 08/06

def par(numero):
    return numero % 2 == 0

numero = int(input("Digite um número inteiro: "))

if par(numero):
    print("O número é par.")
else:
    print("O número é ímpar.")
```

```python
# 3) Busca de Caractere em String - LISTA 08/06

def letra_a(nome):
    return "a" in nome.lower()

nome = input("Digite um nome: ")

if letra_a(nome):
    print("O nome contém a letra 'a'.")
else:
    print("O nome não contém a letra 'a'.")
```

```python
# 4) Verificação de Múltiplo de 5 - LISTA 08/06

def multiplo_de_5(numero):
    return numero % 5 == 0

numero = int(input("Digite um número inteiro: "))

if multiplo_de_5(numero):
    print("O número é múltiplo de 5.")
else:
    print("O número não é múltiplo de 5.")
```

```python
# 5) Classificação de Temperatura Ambiente - LISTA 08/06

def classificar_temp(temp):
    if temp > 30:
        return "Quente"
    else:
        return "Frio"

temperatura = float(input("Digite a temperatura em °C: "))

print("Classificação:", classificar_temp(temperatura))
```

```python
# 6) Controle de Fluxo Interativo (S/N) - LISTA 08/06

def validar_opcao(opcao):
    if opcao == "S":
        return "Continuando..."
    elif opcao == "N":
        return "Saindo..."
    else:
        return "Opção inválida."

opcao = input("Digite S para continuar ou N para sair: ")

print(validar_opcao(opcao))
```

```python
# 7) Triagem Numérica Contextual - LISTA 08/06

def clas_numero(numero):
    if numero > 0:
        return "Positivo"
    elif numero < 0:
        return "Negativo"
    else:
        return "Zero"

numero = float(input("Digite um número: "))

print("Classificação:", clas_numero(numero))
```

```python
# 8) Mapeamento Numérico para Dias da Semana - LISTA 08/06

def dia_semana(numero):
    match numero:
        case 1: return "Segunda-feira"
        case 2: return "Terça-feira"
        case 3: return "Quarta-feira"
        case 4: return "Quinta-feira"
        case 5: return "Sexta-feira"
        case 6: return "Sábado"
        case 7: return "Domingo"
        case _: return "Número inválido"

numero = int(input("Digite um número de 1 a 7: "))

print(dia_semana(numero))
```

```python
# 9) Verificação de Ano Bissexto - LISTA 08/06

def bissexto(ano):
    if ano % 400 == 0:
        return True
    elif ano % 100 == 0:
        return False
    elif ano % 4 == 0:
        return True
    else:
        return False

ano = int(input("Digite um ano: "))

if bissexto(ano):
    print("O ano é bissexto.")
else:
    print("O ano não é bissexto.")
```

```python
# 10) Cálculo de Salário com Horas Extras - LISTA 08/06

def calcular_salario(horas, v_hora):
    if horas <= 40:
        salario = horas * v_hora
    else:
        h_extras = horas - 40
        s_normal = 40 * v_hora
        s_extra = h_extras * (v_hora * 1.5)

        salario = s_normal + s_extra

    return salario


horas = int(input("Digite a quantidade de horas trabalhadas: "))
v_hora = float(input("Digite o valor por hora: "))

print("Salário total: R$", calcular_salario(horas, v_hora))
```

```python
# 11) Soma de Elementos de um Vetor - LISTA 08/06

def soma_vetor(vetor):
    soma = 0
    for numero in vetor:
        soma = soma + numero
    return soma

vetor = []

for i in range(10):
    numero = int(input("Digite um número: "))
    vetor.append(numero)

print("Soma dos valores:", soma_vetor(vetor))
```

```python
# 12) Ordenação Alfabética de Nomes - LISTA 08/06

def ordenar_nomes(nomes):
    nova_lista = []
    for nome in nomes:
        nova_lista.append(nome)
    nova_lista.sort()
    return nova_lista

nomes = []

for i in range(5):
    nome = input("Digite um nome: ")
    nomes.append(nome)

print("Nomes ordenados:", ordenar_nomes(nomes))
```

```python
# 13) Identificação de Extremos em Conjunto de Dados - LISTA 08/06

def maior_menor(vetor):
    vetor.sort()
    return (vetor[-1], vetor[0])

lista = [8, 7, 6, 5, 4, 3, 2, 1]
dados = maior_menor(lista)

print(dados)
```

```python
# 14) Filtragem de Números Pares - LISTA 08/06

def filtrar_pares(lista):
    pares = []
    for numero in lista:
        if numero % 2 == 0:
            pares.append(numero)
    return pares

lista = []

for i in range(10):
    numero = int(input("Digite um número: "))
    lista.append(numero)

print("Números pares:", filtrar_pares(lista))
```

```python
# 15) Cálculo de Média Aritmética - LISTA 08/06
def media_valores(lista):
    if len(lista) == 0:
        return 0
    soma = 0
    for numero in lista:
        soma = soma + numero
    media = soma / len(lista)
    return round(media, 2)

lista = []

for i in range(6):
    numero = float(input("Digite um número: "))
    lista.append(numero)

print("Média:", media_valores(lista))
```

```python
# 16) Contagem Comparativa de Elementos - LISTA 08/06

def contar_maiores(lista, v_ref):
    quantidade = 0
    for numero in lista:
        if numero > v_ref:
            quantidade = quantidade + 1

    return quantidade

lista = []

for i in range(10):
    numero = int(input("Digite um número: "))
    lista.append(numero)

print("Quantidade de números maiores que 50:", contar_maiores(lista, 50))
```

```python
# 17) Inversão de Lista - LISTA 08/06
def inverter_lista(lista):
    nova_lista = []
    for i in range(len(lista)-1, -1, -1):
        nova_lista.append(lista[i])
    return nova_lista

lista = []

for i in range(7):
    numero = int(input("Digite um número: "))
    lista.append(numero)

print("Lista invertida:", inverter_lista(lista))
```

```python
# 18) Identificação de Elementos Duplicados - LISTA 08/06

def encontrar_dupli(lista):
    dupli = []

    for numero in lista:
        if lista.count(numero) > 1 and numero not in dupli:
            dupli.append(numero)
    return dupli

lista = []

for i in range(5):
    numero = int(input("Digite um número: "))
    lista.append(numero)

print("Duplicatas:", encontrar_dupli(lista))
```

```python
# 19) Geração de Lista Aleatória Ordenada - LISTA 08/06

import random
def gerar_lista_aleatoria(qtd,inicio,fim):
    lista = []
    for i in range(qtd):
        num = random.randint(inicio,fim)
        lista.append(num)
    #lista.sort() #Mega Sena
    return lista
print(gerar_lista_aleatoria(10,1,100))
```

```python
# 20) Verificação de Existência em Lista - LISTA 08/06

def verificar_valor(lista,numero):
    for i in range(len(lista)):
        if numero == lista[i]:
            return True
        elif i == len(lista) - 1:
            return False

lista = gerar_lista_aleatoria(6,1,10)
print(lista)

valor = int(input('Digite um valor: '))
print(f'Valor encontrado = {verificar_valor(lista,valor)}')
```
