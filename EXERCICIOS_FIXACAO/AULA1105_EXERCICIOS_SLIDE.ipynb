```python
# ATIVIDADE 1 (SLIDE) - DICIONÁRIOS

produto = {
    "nome": "Mouse",
    "preco": 89.90,
    "estoque": 46,
    "categoria": "Informática"
}

for chave in produto:
    print(chave, ":", produto[chave])

produto["preco"] = 120

produto["desconto"] = 10

print()

for chave in produto:
    print(chave, ":", produto[chave])
```

```python
# ATIVIDADE 2 (SLIDE) - DICIONÁRIOS

alunos = [
    {"nome": "João", "media": 8},
    {"nome": "Maria", "media": 6},
    {"nome": "Doces", "media": 9}
]

for aluno in alunos:
    print("\nAluno:", aluno["nome"])
    print("Média:", aluno["media"])

print()

print("Alunos acima da média:")

for aluno in alunos:

    if aluno["media"] > 7:
        print(aluno["nome"])
```

```python
# ATIVIDADE 3 (SLIDE) - DICIONÁRIOS

produto = {
    "nome": "garrafinha de água",
    "preco": 50,
    "quantidade": 365
}

produto["valor_total"] = produto["preco"] * produto["quantidade"]

for chave in produto:
    print(chave, ":", produto[chave])
```

```python
# ATIVIDADE 1 (SLIDE) - CONJUNTOS

inscritos = ["Ana", "Bruno", "Ana", "Carla", "Bruno", "Diego"]

unicos = set(inscritos)

print(unicos)

print("Quantidade de pessoas únicas:", len(unicos))
```

```python
# ATIVIDADE 2 (SLIDE) - CONJUNTOS

turma_python = {"Ana", "Bruno", "Carla", "Diego"}
turma_java = {"Carla", "Diego", "Elena", "Fábio"}

print("Nas duas turmas:")
print(turma_python & turma_java)

print()

print("Em pelo menos uma turma:")
print(turma_python | turma_java)

print()

print("Só em Python:")
print(turma_python - turma_java)
```

```python
# ATIVIDADE 3 (SLIDE) - CONJUNTOS

notas = [7, 8, 7, 9, 10, 8, 6, 10]

notas_unicas = set(notas)

lista_ordenada = sorted(notas_unicas)

print(lista_ordenada)

print("Quantidade de notas únicas:", len(lista_ordenada))
```

```python
# ATIVIDADE FINAL (SLIDE) "DESAFIO FINAL"

turma = [
    {
        "nome": "Joaquim",
        "notas": [8, 7, 9]
    },

    {
        "nome": "Manuel",
        "notas": [5, 6, 4]
    },

    {
        "nome": "Macedo",
        "notas": [10, 8, 9]
    }
]

aprovados = set()

for aluno in turma:

    notas = aluno["notas"]

    media = sum(notas) / len(notas)

    print("Aluno:", aluno["nome"])
    print(f"Média: {media:.2f}")

    if media >= 7:
        print("Aprovado")

        aprovados.add(aluno["nome"])

    print()

print("Nomes dos aprovados:")
print(aprovados)
```
