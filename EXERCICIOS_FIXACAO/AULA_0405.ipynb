```python
# ATIVIDADE 8 (LISTA DE EXERCÍCIOS)

poderes = []
while True:
    poder = input('Digite o novo poder: ')
    if poder.lower() == 'sair':
        print(poderes)
        break
    else:
        poderes.append(poder.capitalize())
        print(f'Poder - {poder} foi adicionado a lista!')
```
