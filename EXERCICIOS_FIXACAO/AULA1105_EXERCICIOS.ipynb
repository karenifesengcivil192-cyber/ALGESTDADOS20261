```python
# ATIVIDADE 1

coordenadas = ((10, 20), (3, 4), (12, 18))

for ponto in coordenadas:

    print("X =", ponto[0])
    print("Y =", ponto[1])
    print()

soma = 0

maior = coordenadas[0][0]

for ponto in coordenadas:

    soma += ponto[0]
    soma += ponto[1]

    if ponto[0] > maior:
        maior = ponto[0]

    if ponto[1] > maior:
        maior = ponto[1]

print("Soma de todos os valores:", soma)
print("Maior valor:", maior)
```

```python
# ATIVIDADE 2

produtos = set()

for i in range(1, 6):

    produto = input(f"Digite o {i}° produto: ")

    produtos.add(produto)

print("\nProdutos cadastrados:")

for p in produtos:
    print(p)

print("\nQuantidade de produtos únicos:", len(produtos))
```

```python
# ATIVIDADE 3

numeros = (2, 5, 8, 7, 10, 3)

pares = 0
soma_impares = 0

menor = numeros[0]

for n in numeros:

    if n % 2 == 0:
        pares += 1

    else:
        soma_impares += n

    if n < menor:
        menor = n

print("Quantidade de pares:", pares)
print("Soma dos ímpares:", soma_impares)
print("Menor valor:", menor)
```

```python
# ATIVIDADE 7

notas = (7.5, 8.0, 6.5, 9.0)
soma = 0

maior = notas[0]
menor = notas[0]

for n in notas:
    soma += n

    if n > maior:
        maior = n
    if n < menor:
        menor = n

media = soma / len(notas)

print(f"Média: {media:.2f}")
print("\nMaior nota:", maior)
print("Menor nota:", menor)

print()
print("Notas armazenadas:")

for n in notas:
    print(n)
```
