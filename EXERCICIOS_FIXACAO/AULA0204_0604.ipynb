```python
#ATIVIDADE 5
num = int(input("Digite um número: "))

if num >= 0:
    conta = 0

    for i in range(0, num + 1, 2):
        print(i)
        conta += 1

    print(f"tem, {conta} pares")

else:
    print("Número não positivo")
```

```python
#ATIVIDADE 6
n = int(input("Digite a quantidade de alunos: "))

if n > 0:
    soma = 0

    for i in range(1, n+1):
        nota = float(input(f"Digite a nota do {i}° aluno:"))

        if 0 <= nota  <= 10:
            soma += nota
        else:
            print("Nota inválida")

    media = soma / n
    print(f"A média de notas é: {media:.2f}")

else:
    print("Quantidade inválida")
```

```python
#ATIVIDADE 7
n = int(input("Digite um número (0 a 12): "))

if 0 <= n <= 12:
    fatorial = 1

    for i in range(1, n+1):
        fatorial *= i

    print(f"O fatorial de {n} é: {fatorial}")

else:
    print("Valor inválido")
```

```python
#ATIVIDADE 8
n = int(input("Digite a quantidade de valores (mínimo 2): "))

if n >= 2:
    for i in range(1, n+1):
        valor = float(input(f"Digite o {i}° valor: "))

        if i == 1:
            maior_valor = valor
            menor_valor = valor
        else:
            if valor > maior_valor:
                maior_valor = valor
            if valor < menor_valor:
                menor_valor = valor

    print(f"\nMaior valor: {maior_valor}")
    print(f"Menor valor: {menor_valor}")

else:
    print("Quantidade inválida")
```

```python
#ATIVIDADE 11
n = int(input("Digite a quantidade de números: "))

if n > 0:
    positivos = 0
    negativos = 0
    zeros = 0

    for i in range(1, n+1):
        numero = int(input(f"Digite o {i}° número: "))

        if numero > 0:
            print("Positivo")
            positivos += 1
        elif numero < 0:
            print("Negativo")
            negativos += 1
        else:
            print("Zero")
            zeros += 1

    print(f"Quantidade de positivos: {positivos}")
    print(f"Quantidade de negativos: {negativos}")
    print(f"Quantidade de zeros: {zeros}")

else:
    print("Quantidade inválida")
```

```python
#ATIVIDADE 12
a = int(input("Digite o valor de A: "))
b = int(input("Digite o valor de B: "))

if a <= b:
    soma = 0
    contador = 0

    for i in range(a, b+1):
        if i % 2 != 0:
            soma += i
            contador += 1
    print(f"Quantidade de ímpares: {contador}")
    print(f"Soma dos ímpares: {soma}")

else:
    print("Intervalo inválido")
```

```python
#ATIVIDADE 13
n = int(input("Digite um número (maior que 1): "))
eh_primo = True

for i in range(2, n):
    if n % i == 0:
        eh_primo = False
        break

if eh_primo:
    print("É primo")
else:
    print("Não é primo")
```

```python
#ATIVIDADE 15
nome = input("Digite o nome do aluno: ")
n = int(input("Digite a quantidade de notas (mínimo 2): "))
soma = 0

for i in range(n):
    nota = float(input(f"Digite a nota {i+1}: "))

    if nota < 0 or nota > 10:
        print("Nota inválida! Considerando 0.")
        nota = 0

    soma += nota
media = soma / n

if media >= 9:
    conceito = "A"
elif media >= 7:
    conceito = "B"
elif media >= 5:
    conceito = "C"
elif media >= 3:
    conceito = "D"
else:
    conceito = "E"

print("\nAluno:", nome)
print(f"Média: {media:.2f}")
print("Conceito:", conceito)
```
