```python
#EXEMPLOS DO SLIDE
senha_correta = "python2026"
tentativas = 0
max_tentativas = 3

while tentativas < max_tentativas:
    senha = input("Digite a senha: ")
    tentativas += 1

    if senha == senha_correta:
        print("Acesso liberado! Bem-vindo!")
        break
    else:
        restam = max_tentativas - tentativas
        if restam > 0:
            print(f"Senha incorreta! {restam} restante(s).")

if tentativas == max_tentativas:
    if senha != senha_correta:
        print("Conta bloqueada! Muitas tentativas.")
```

```python
#ATIVIDADE 12 DOS EXERCÍCIOS 09/04
palavra = input('Digite um palavra: ')
inicio = 0
fim = len(palavra)-1
meio = fim/2
while True:
    if palavra[inicio].upper()!=palavra[fim].upper():
        print('Não é palindromo')
        break
    elif inicio == meio or fim == meio:
        print(f'A {inicio+1}ª letra é {palavra[inicio]}')
        print('É palíndromo')
        break
    else:
        print(f'A {inicio+1}ª letra é {palavra[inicio]} e A {fim+1}ª letra é {palavra[fim]}')
        inicio+=1
        fim-=1
```
